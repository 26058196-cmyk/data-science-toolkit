```python
#Part 1-Google Colab and Python
#1. Create a new notebook and code the following

#1. Create a numpy array of size 2 x 2
import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])
print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4

random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes

ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of ones")
print(ones_array)

print("Array of zeroes")
print(zeros_array)

```

```python
#4. Create a random numpy array of size 6x3 and transpose it

random6x3 = np.random.rand(6x3)
transposed_random6x3 = random6x3.T

print("Random 6x3 array")
print(random6x3)

print("Transposed 6x3 array")
print(transposed_random6x3)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product

n,p,m = 2,3,4

randomnxp = np.random.rand(n,p)
randompxm = np.random.rand(p,m)

dot_product_result = np.dot(randomnxp, randompxm)

print("Random n x p array")
print(randomnxp)

print("Random p x m array")
print(randompxm)

print("Dot product result")
print(dot_product_result)


```
